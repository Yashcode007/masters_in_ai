```python
import torch
from typing import Any
```

```python
class MyModel(torch.nn.Module):
  def __init__(self , layer_size=[512,512,512]) -> None:
    super().__init__()
    layers = []
    layers.append(torch.nn.Flatten())
    c = 128*128*3
    for s in layer_size:
      layers.append(torch.nn.Linear(c,s))
      layers.append(torch.nn.ReLU())
      c = s
    layers.append(torch.nn.Linear(c,102))
    self.model = torch.nn.Sequential(*layers)

  def forward(self,x) -> Any:
    return self.model(x)
```

```python
net = MyModel()
x = torch.randn(10,3,128,128)
net(x)
```

We can do this for one layer network , or two layer network or so on

```python
x = torch.randn(10,3,128,128)
net0 = MyModel([])
print(f"net0:{net0(x)}")
net1 = MyModel([512])
print(f"net1:{net1(x)}")
net2 = MyModel([512,512])
print(f"net2:{net2(x)}")
```

As we can see that the values are getting smaller and smaller . This signifies vanishing and exploding problem .

Now , lets apply normalisation to a set of these examples, to see how values are changing , etc

```python
x = torch.randn(10,3,128,128)
for n in range(15):
  netn = MyModel([512]*n)
  print(f"{n=} {netn(x).norm()=}")
```

As we go down we can see the activations are getting smaller and smaller . But after some point of time , they stop to get smaller . This is because of the bias in the normalisation

Lets try this without the bias

```python
class MyModelnobias(torch.nn.Module):
  def __init__(self , layer_size=[512,512,512]) -> None:
    super().__init__()
    layers = []
    layers.append(torch.nn.Flatten())
    c = 128*128*3
    for s in layer_size:
      layers.append(torch.nn.Linear(c,s, bias=False))
      layers.append(torch.nn.ReLU())
      c = s
    layers.append(torch.nn.Linear(c,102, bias=False))
    self.model = torch.nn.Sequential(*layers)

  def forward(self,x) -> Any:
    return self.model(x)
```

```python
x = torch.randn(10,3,128,128)
for n in range(15):
  netn = MyModelnobias([512]*n)
  print(f"{n=} {netn(x).norm()=}")
```

Now you can see that the activations are getting smaller and smaller and smaller . This is the concept of Vanishing activations

Now, let us try to solve the following scenario with the help of batch normalisation . We will be adding a normalisation just before the ReLu and after the Linear Layer

```python
class MyModelBatchNorm(torch.nn.Module):
  def __init__(self , layer_size=[512,512,512]) -> None:
    super().__init__()
    layers = []
    layers.append(torch.nn.Flatten())
    c = 128*128*3
    for s in layer_size:
      layers.append(torch.nn.Linear(c,s, bias=False))
      layers.append(torch.nn.BatchNorm1d(s))
      layers.append(torch.nn.ReLU())
      c = s
    layers.append(torch.nn.Linear(c,102, bias=False))
    self.model = torch.nn.Sequential(*layers)

  def forward(self,x) -> Any:
    return self.model(x)
```

```python
x = torch.randn(10,3,128,128)
for n in range(15):
  netn = MyModelBatchNorm([512]*n)
  print(f"{n=} {netn(x).norm()=}")
```

Now you can see , the activation stays high(or consistent). We dont see any vanishing gradient over here .

The same illustration we can do for the Layer Norm

```python
class MyModelLayerNorm(torch.nn.Module):
  def __init__(self , layer_size=[512,512,512]) -> None:
    super().__init__()
    layers = []
    layers.append(torch.nn.Flatten())
    c = 128*128*3
    for s in layer_size:
      layers.append(torch.nn.Linear(c,s, bias=False))
      layers.append(torch.nn.LayerNorm(s))
      layers.append(torch.nn.ReLU())
      c = s
    layers.append(torch.nn.Linear(c,102, bias=False))
    self.model = torch.nn.Sequential(*layers)

  def forward(self,x) -> Any:
    return self.model(x)
```

```python
x = torch.randn(10,3,128,128)
for n in range(15):
  netn = MyModelLayerNorm([512]*n)
  print(f"{n=} {netn(x).norm()=}")
```

We can still see the gradients to be consistent.

We can add normalisation before the linear layer also . We should always add the normalisation or the bias layer before the ReLu , as many negatives becomes positive after norm , so less values will be provided to the Relu to make the data zero .

```python
class MyModelLayerNorm_2(torch.nn.Module):
  def __init__(self , layer_size=[512,512,512]) -> None:
    super().__init__()
    layers = []
    layers.append(torch.nn.Flatten())
    c = 128*128*3
    for s in layer_size:
      layers.append(torch.nn.LayerNorm(s, bias = False))
      layers.append(torch.nn.Linear(c,s))
      layers.append(torch.nn.ReLU())
      c = s
    layers.append(torch.nn.Linear(c,102, bias=False))
    self.model = torch.nn.Sequential(*layers)

  def forward(self,x) -> Any:
    return self.model(x)
```

```python
x = torch.randn(10,3,128,128)
for n in range(15):
  netn = MyModelLayerNorm_2([512]*n)
  print(f"{n=} {netn(x).norm()=}")
```
