```python
import torch
from typing import Any
```

```python
class Net(torch.nn.Module):
  def __init__(self)-> None:
    super().__init__()
    self.fc1 = torch.nn.Linear(1,10)
    self.fc2 = torch.nn.Linear(10,1)
    self.act = torch.nn.ReLU()

  def forward(self,x: torch.Tensor) -> Any:
    x = self.fc1(x)
    x = self.act(x)
    x = self.fc2(x)
    return x
```

Always and always our first line of init code is going to be calling the super init, so that the pytorch can sort of start wrapping all the other stuff we do appropiately .
We should always define the modules , variation or infact the activation function inside the init

```python
net = Net()
net(torch.tensor([1.0]))
```

```python
class Net2(torch.nn.Module):
  def __init__(self)->None:
    super().__init__()
    self.fc1_x = torch.nn.Linear(1,10)
    self.fc1_y = torch.nn.Linear(1,10)
    self.fc2 = torch.nn.Linear(10,1)
    self.act = torch.nn.ReLU()

  def forward(self,x:torch.Tensor, y:torch.Tensor)-> torch.Tensor:
    x = self.fc1_x(x)
    x = self.act(x)
    y = self.fc1_y(y)
    y = self.act(y)
    return self.fc2(x+y)
```

```python
net2 = Net2()
net2(torch.Tensor([1.0]), torch.Tensor([2.0]))
```
