```python
import torch
```

```python
model = torch.nn.Linear(10,1)
```

```python
x = torch.randn(20,10)
y = torch.randn(20,1)
```

```python
print(f'{x=} {y=}')
```

```python
pred_y = model(x)
print(pred_y)
```

```python
# Now we need a way to compare y and pred_y
loss = torch.nn.functional.mse_loss(pred_y,y)
```

```python
print(loss)
```

```python
def mse_loss(pred_y,y):
  return torch.mean((pred_y-y) ** 2)
mse_loss(pred_y,y)
```
